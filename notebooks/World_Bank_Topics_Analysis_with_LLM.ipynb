```python
from transformers import AutoConfig, AutoModel, AutoTokenizer
import transformers
import datasets


model_path = 'microsoft/deberta-v3-small'
tokenizer = AutoTokenizer.from_pretrained(model_path)

data = datasets.load_dataset("avsolatorio/doc-topics-synthetic_data", data_dir="processed", split="train", revision="e6d841658894fc096d8e0df3fb0567c201250bfc")

classes = list(set(wb_topics["all_topics"]))
class2id = {class_:id for id, class_ in enumerate(classes)}
id2class = {id:class_ for class_, id in class2id.items()}
```

```python
def preprocess_function(example):
    try:
        example["valid"] = True

        text = example["sentence"] or ""

        all_labels = example["topics"] or []

        if text == "" or len(all_labels) == 0:
            example["valid"] = False

        labels = [0. for i in range(len(classes))]
        for label in all_labels:
            if label not in class2id:
                example["valid"] = False
                continue

            label_id = class2id[label]
            labels[label_id] = 1.

        if not any(labels):
            example["valid"] = False

        example = tokenizer(text, truncation=True)
        example['labels'] = labels
        return example
    except Exception as e:
        print(example)
        raise e

dataset = data
print(len(dataset))

tokenized_dataset = dataset.map(preprocess_function)
tokenized_dataset = tokenized_dataset.filter(lambda x: x["valid"])

tokenized_dataset
```

```python
from transformers import DataCollatorWithPadding


data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

```python
import evaluate
import numpy as np


clf_metrics = evaluate.combine(["accuracy", "f1", "precision", "recall"])


def sigmoid(x):
    return 1/(1 + np.exp(-x))


def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = sigmoid(predictions)
    predictions = (predictions > 0.5).astype(int).reshape(-1)

    return clf_metrics.compute(predictions=predictions, references=labels.astype(int).reshape(-1))
```

```python
from transformers import AutoModelForSequenceClassification, TrainingArguments, Trainer, EarlyStoppingCallback

def model_init():
    return AutoModelForSequenceClassification.from_pretrained(
        model_path,
        num_labels=len(classes),
        id2label=id2class,
        label2id=class2id,
        problem_type="multi_label_classification"
    )

model = model_init()
```

```python
# Split the data into 5 groups
# Train 5 separate models
# Use the 4 models to validate the LLM assigned topics as well as infer potentially missing topics.
# A label is valid if 3 out of 4 models agree on a topic for a given sample.
# Once all the data groups have been imputed, shuffle and repeat the process until convergence.
```

```python
from datasets import Dataset

# Get the length of the dataset
dataset_length = len(tokenized_dataset)

# Calculate the size of each subset
subset_size = dataset_length // 5

# Split into 5 equally sized subdatasets
subdatasets = []
remaining_dataset = tokenized_dataset

for i in range(4):  # Split the first 4 subdatasets
    split_dataset = remaining_dataset.train_test_split(test_size=subset_size, seed=42)
    subdatasets.append(split_dataset['test'])
    remaining_dataset = split_dataset['train']

# Append the remaining part as the fifth subdataset
subdatasets.append(remaining_dataset)

# subdatasets is now a list of 5 subdatasets
len(subdatasets)
```
